# Box-constrained trials

This notebook is the box-constrained counterpart of `main.ipynb`. It runs projected NTRQN and SciPy L-BFGS-B on selected bound-constrained CUTEst problems.

```python
from data.CUTEst.check_CUTEst_problems import problemsToRun
from qnlab.experiment.trial import trial
from qnlab.problem import CUTEstQNProblem
from qnlab.util.method import get_box_methods

methods, *_ = get_box_methods(m=10, MI=1500)
```

```python
precision = 64

# Keep this list short for interactive inspection. Add any names returned by
# problemsToRun(precision, constraints="bound") as needed.
problems = [
    "3PK",
    "BIGGSB1",
    "CAMEL6",
    "GENROSEB",
    "HART6",
    "HS1",
    "HS38",
]

valid_problems = set(problemsToRun(precision, constraints="bound"))

for name in problems:
    if name not in valid_problems:
        print(f"Skipping {name}: unavailable at precision {precision}.")
        continue

    prob = CUTEstQNProblem(name, precision)
    print(prob.bounds)
    try:
        trial(prob, name, methods, bounds=prob.bounds)
    except Exception as e:  # noqa: BLE001 - continue the experiment batch
        print(f"Problem {name} failed with exception: {e}")

print("Done.")
```
